```python
from langchain_text_splitters import RecursiveCharacterTextSplitter , Language
```

```python
splitter = RecursiveCharacterTextSplitter.from_language(
    language = Language.PYTHON,
    chunk_size = 40,
    chunk_overlap = 0
)
```

```python
code = """

def add(a,b):
    return a+b

def sub(a,b):
    return a-b

def product(a,b):
    return a*b
    
for i in range(100):
    print(i)
    """
```

```python
chunks = splitter.split_text(code)
```

```python
for i in chunks:
    print(i)
    print(len(i))
    print("-"*30)
```
